```python
class stack:

    def __init__(self):
        self.data=UnsortedPriorityQueue()
        self.cont=0
    def push(self,element):
        self.data.add(self.cont,element)
        self.cont-=1

    def top(self):
        self.data.min()[1]

    def pop(self):
        if len(self.data)==0:
            raise Empty("El stack esta vacio")
        return self.data.remove_min()[1]
```
