##maxiomo y minomo

```python
def MinMax(L):
    if  len(L)==1:
        minim=L[0]
        maxim=L[0]
        return minim,maxim
    else:
        maxim,minim=MinMax(L[1:])
        if L[0]>maxim:
            maxim=L[0]
        if L[0]<minim:
            minim=L[0]
    return maxim,minim
```

cuente el numero de picos en la lista L

```python
def picos(L):
    if len(L)==3:
        if L[0]<L[1] and L[1]>L[2]:
            return 1
        return 0
    else:
        if L[0]<L[1] and L[1]>L[2]:
            return 1+picos(L[1:])
    return picos(L[1:])
```

```python
class Empty(Exception):
    """Excepción personalizada para cuando la pila está vacía."""
    pass

class ArrayStack:
    def __init__(self):
        """Crea una pila vacía."""
        self._data = []  # Instancia de lista no pública

    def __len__(self):
        """Devuelve el número de elementos en la pila."""
        return len(self._data)

    def is_empty(self):
        """Devuelve True si la pila está vacía."""
        return len(self._data) == 0

    def push(self, e):
        """Añade el elemento 'e' al tope de la pila."""
        self._data.append(e)  # El nuevo elemento se almacena al final de la lista

    def top(self):
        """Devuelve (pero no elimina) el elemento en el tope de la pila.
        
        Lanza la excepción Empty si la pila está vacía.
        """
        if self.is_empty():
            raise Empty('Stack is empty')
        return self._data[-1]  # El último elemento de la lista

    def pop(self):
        """Elimina y devuelve el elemento del tope de la pila (LIFO).
        
        Lanza la excepción Empty si la pila está vacía.
        """
        if self.is_empty():
            raise Empty('Stack is empty')
        return self._data.pop()  # Elimina el último elemento de la lista
```

```python
def Contar(S, S1, a):
    contador=0
    while not S.is_empty():
        x=S.pop()
        if x==a:
            contador+=1
    while not S1.is_empty():
        x=S.pop()
        if x==a:
            contador+=1
    return contador
```
